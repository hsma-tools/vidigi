## What happens when the queue exceeds the available space? 

Here, we override the arrival rate so that the model will quickly exceed our 'step_snapshot_limit' parameter. 

By default, we will then receive a '+ x more' text at the end of our queue, indicating the extra number of waiting individuals. 

````python
g.arrival_rate = 2

my_trial = Trial()

my_trial.run_trial()
````

````python
animate_activity_log(
        event_log=my_trial.all_event_logs[my_trial.all_event_logs['run']==1],
        event_position_df= event_position_df,
        entity_col_name="patient",
        scenario=g(),
        debug_mode=True,
        setup_mode=False,
        every_x_time_units=1,
        include_play_button=True,
        resource_icon_size=15,
        text_size=20,
        entity_icon_size=13,
        gap_between_entities=6,
        gap_between_queue_rows=25,
        gap_between_resource_rows=25,
        plotly_height=600,
        frame_duration=200,
        plotly_width=1000,
        override_x_max=300,
        override_y_max=500,
        limit_duration=g.sim_duration,
        wrap_queues_at=25,
        step_snapshot_max=125,
        time_display_units="dhm",
        display_stage_labels=False,
        add_background_image="https://raw.githubusercontent.com/Bergam0t/vidigi/refs/heads/main/examples/example_1_simplest_case/Simplest%20Model%20Background%20Image%20-%20Horizontal%20Layout.drawio.png",
        step_snapshot_limit_gauges=False
    )
````

If we instead set `step_snapshot_limit_gauges=True`, we will instead display a bar that fills up relative to the maximum observed number of entities in that queue (above and beyond those displayed as icons). 

This works best if paired with a `frame_transition_duration` of 30 (almost invisible).

````python
animate_activity_log(
        event_log=my_trial.all_event_logs[my_trial.all_event_logs['run']==1],
        event_position_df= event_position_df,
        entity_col_name="patient",
        scenario=g(),
        debug_mode=True,
        setup_mode=False,
        every_x_time_units=1,
        include_play_button=True,
        resource_icon_size=15,
        text_size=20,
        entity_icon_size=13,
        gap_between_entities=6,
        gap_between_queue_rows=25,
        gap_between_resource_rows=25,
        plotly_height=600,
        frame_duration=200,
        # frame_transition_duration=0,
        plotly_width=1000,
        override_x_max=300,
        override_y_max=500,
        limit_duration=g.sim_duration,
        wrap_queues_at=25,
        step_snapshot_max=125,
        time_display_units="dhm",
        display_stage_labels=False,
        add_background_image="https://raw.githubusercontent.com/Bergam0t/vidigi/refs/heads/main/examples/example_1_simplest_case/Simplest%20Model%20Background%20Image%20-%20Horizontal%20Layout.drawio.png",
        step_snapshot_limit_gauges=True,
    )
````
